# Task 1 — Scatter: Life Expectancy Change Over Time
## GDP per Capita vs Life Expectancy — Asia (2002 vs 2007)

```python
import pandas as pd
import plotly.express as px

df = px.data.gapminder()
print(f"Loaded: {len(df)} rows")
```

```python
# Filter: Asia, years 2002 and 2007
asia = df[(df['continent'] == 'Asia') & (df['year'].isin([2002, 2007]))].copy()
asia['year'] = asia['year'].astype(str)  # treat year as category for discrete colour

# Build scatter chart
fig = px.scatter(
    data_frame=asia,
    x='gdpPercap',
    y='lifeExp',
    color='year',
    hover_name='country',
    hover_data={'year': False},
    title='Asian nations gained in wealth and longevity — most shifted right and up (2002 → 2007)',
    labels={
        'gdpPercap': 'GDP per Capita (log scale, US$)',
        'lifeExp': 'Life Expectancy (years)'
    },
    color_discrete_map={'2002': '#AAAAAA', '2007': '#2E75B6'},
    log_x=True,
    height=520
)

fig.update_traces(marker=dict(size=9, opacity=0.85))

fig.update_layout(
    plot_bgcolor='white',
    paper_bgcolor='white',
    font=dict(family='Arial', size=12),
    legend=dict(title='Year', orientation='h', y=1.08),
    margin=dict(l=60, r=40, t=80, b=50),
)

fig.update_xaxes(gridcolor="#E04949", showline=True, linecolor='#CCCCCC')
fig.update_yaxes(gridcolor='#EEEEEE', showline=True, linecolor='#CCCCCC')

fig.show()
```
